```python
'''1. E-commerce Discount Calculator

Scenario: An online store offers discounts based on the purchase amount:

10% discount for purchases between $100 and $500

20% discount for purchases above $500

No discount for purchases below $100


Task: Write a program that takes the purchase amount as input and calculates the discount and final amount to be paid.'''

Amount = float(input("Enter purchase amount: $"))

if Amount < 100:
    discount = 0
elif Amount <= 500:
    discount = Amount * 0.10
else:
    discount = Amount * 0.20

final_amount = Amount - discount

print("Purchase Amount: $", Amount)
print("Discount: $", discount)
print("Final Amount to Pay: $", final_amount)
```

```python
'''6. Electricity Bill Calculator

Scenario: An electricity company charges its customers as follows:

First 100 units: $0.5 per unit

Next 100 units (101-200): $0.75 per unit

Above 200 units: $1 per unit


Task: Write a program that accepts the number of units consumed and calculates the total bill.'''

units = int(input("Enter the units of Electricity consumed:"))

if units <= 100:
    bill = units * 0.5
elif units <= 200:
    bill = (100 * 0.5) + (units - 100) * 0.75
else:
    bill = (100 * 0.5) + (100 * 0.75) + (units - 200) * 1

print("Total Electricity Bill: $", bill)
```

```python
'''14. Password Strength Checker
Accept a password. If its length is at least 8 characters, check whether it contains a digit. 
If it does, check whether it contains an uppercase letter. 
Classify it as strong or moderate based on these conditions. 
Otherwise, classify it as weak.'''

password = input("Enter your password: ")

if len(password) >= 8:
    if any(char.isdigit() for char in password):
        if any(char.isupper() for char in password):
            print("Strong Password")
        else:
            print("Moderate Password")
    else:
        print("Moderate Password")
else:
    print("Weak Password")
```

```python
'''20. Mini Banking System
Accept account type, account balance, and transaction type (deposit or withdrawal). 
If the account type is savings, check the transaction type. For withdrawals, check minimum balance requirements; 
for deposits, validate the deposit amount. 
If the account type is current, apply a different minimum balance rule.'''

account = input("Enter account type (savings/current): ").lower()
balance = float(input("Enter account balance: ₹"))
transaction = input("Enter transaction (deposit/withdrawal): ").lower()
amount = float(input("Enter transaction amount: ₹"))

if account == "savings":
    if transaction == "withdrawal":
        if balance - amount >= 1000:
            balance = balance - amount
            print("Withdrawal Successful")
        else:
            print("Minimum balance of ₹1000 must be maintained")
    else:
        if transaction == "deposit":
            if amount > 0:
                balance = balance + amount
                print("Deposit Successful")
            else:
                print("Invalid deposit amount")
        else:
            print("Invalid transaction type")

elif account == "current":
    if transaction == "withdrawal":
        if balance - amount >= 5000:
            balance = balance - amount
            print("Withdrawal Successful")
        else:
            print("Minimum balance of ₹5000 must be maintained")
    else:
        if transaction == "deposit":
            if amount > 0:
                balance = balance + amount
                print("Deposit Successful")
            else:
                print("Invalid deposit amount")
        else:
            print("Invalid transaction type")
else:
    print("Invalid account type")

print("Available Balance: ₹", balance)
```
